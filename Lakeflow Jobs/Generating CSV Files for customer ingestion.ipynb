```sql
%python
catalog_base = "databricks_simulated_retail_customer_data"
schema_base = "v01"
table = "customers"

catalog_destino = "new_sample"
schema_destino = "retail_data"
volume = "customer_data"

table_fqn = f"{catalog_base}.{schema_base}.{table}"
output_dir = f"/Volumes/{catalog_destino}/{schema_destino}/{volume}/{table}_first_1000_csv"

df = spark.table(table_fqn).limit(1000)

(df.coalesce(1)
   .write
   .mode("overwrite")
   .option("header", "true")
   .csv(output_dir))

print(f"CSV salvo em: {output_dir}")
```

```sql
%python
catalog_base = "databricks_simulated_retail_customer_data"
schema_base = "v01"
table = "customers"

catalog_destino = "new_sample"
schema_destino = "retail_data"
volume = "customer_data"

table_fqn = f"{catalog_base}.{schema_base}.{table}"
output_dir = f"/Volumes/{catalog_destino}/{schema_destino}/{volume}/{table}_second_1000_csv"

df = spark.sql(f"""
SELECT *
FROM {table_fqn}
LIMIT 1000 OFFSET 1000
""")

(df.coalesce(1)
   .write
   .mode("overwrite")
   .option("header", "true")
   .csv(output_dir))

print(f"CSV salvo em: {output_dir}")
```
