```python
%sql
drop table if exists new_sample.retail_data.customer_orders_silver;

create or replace table new_sample.retail_data.customer_orders_silver
as
select 
    c.customer_id,
    c.customer_name,
    c.state,
    c.city,
    o.order_number,
    o.order_datetime,
    o.number_of_line_items,
    o.promo_info
from 
    new_sample.retail_data.customer_bronze as c
left join
    new_sample.retail_data.orders_bronze as o
on
    c.customer_id = o.customer_id;
```
