```python
%sql
drop table if exists new_sample.retail_data.customer_sales_silver;

create or replace table new_sample.retail_data.customer_sales_silver
as
select 
    c.customer_id,
    c.customer_name,
    c.loyalty_segment,
    c.units_purchased,
    s.product_name,
    s.order_date,
    s.total_price
from 
    new_sample.retail_data.customer_bronze as c
left join
    new_sample.retail_data.sales_bronze as s
on
    c.customer_id = s.customer_id;
```

```python
%python
df = spark.sql("""
               SELECT * FROM new_sample.retail_data.customer_sales_silver
               """)

## Check for duplicate records
duplicate_exists = df.count() >= df.dropDuplicates().count()

if duplicate_exists:
  print("Duplicate records found in the table")
else:
  print("No duplicate records found in the table")

## Set boolean flag in task values
dbutils.jobs.taskValues.set(key="has_duplicates", value=duplicate_exists)

```
