```python
%sql
CREATE SCHEMA IF NOT EXISTS new_sample.pipeline_retail_data
COMMENT 'this schema will be used to storage Files and tables used in the our Spark declarative pipeline';
```

```python
%sql
ALTER VOLUME new_sample.pipeline_retail_data.orders
RENAME TO new_sample.pipeline_retail_data.Retail_Data_JSON_Files;
```

```python
%sql
-- Creating a Bronze Straming table from raw JSON File in your source volume
CREATE OR REFRESH STREAMING TABLE new_sample.pipeline_retail_data.Orders_Bronze
AS 
SELECT 
*,
current_timestamp() as processing_time,
_metadata.file_name as source_file
FROM STREAM read_files(
                        "${source}/orders",
                        format => "json"
                        )
;
```
